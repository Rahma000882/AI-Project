# **Unsupervised Learning**
**Unsupervised learning is a type of machine learning where the model is trained on unlabeled data without any guidance or supervision.**

**Unlike supervised learning, where the algorithm is provided with labeled data (input-output pairs) to learn from, unsupervised learning algorithms must discover the patterns and structures in the input data by themselves.**

**The main goal of unsupervised learning is to find hidden patterns or intrinsic structures within the data.**

**This can involve tasks such as *clustering*, where the algorithm groups similar data points together, or *dimensionality reduction*, where the algorithm reduces the number of features while preserving the essential information.**

**Unsupervised learning models identify patterns through various techniques that are designed to analyze the structure of the input data. Here are some common methods:**

1. **Clustering:**
- **Clustering algorithms group similar data points together based on certain criteria, such as distance or density.**
- **The model identifies clusters of data points that are closer to each other in feature space compared to others.**
- **Examples of clustering algorithms include k-means clustering, hierarchical clustering, and DBSCAN.**


2. **Dimensionality Reduction:**
- **Dimensionality reduction techniques aim to capture the essential information in the data while reducing its dimensionality.**
- **Principal Component Analysis (PCA) and t-distributed Stochastic Neighbor Embedding (t-SNE) are examples of dimensionality reduction methods.**
- **These techniques identify patterns by finding lower-dimensional representations of the data that preserve its structure as much as possible.**


3. **Density Estimation:**
- **Density estimation methods model the probability distribution of the data. They identify patterns by estimating the underlying distribution of the data points.**
- **Kernel Density Estimation (KDE) and Gaussian Mixture Models (GMMs) are examples of density estimation techniques used in unsupervised learning.**


4. **Association Rule Mining:**
- **Association rule mining identifies relationships or associations between variables in the data.**
- **It discovers patterns by finding frequent co-occurrences of items or events in the dataset.**
- **Apriori and FP-growth are common algorithms used for association rule mining.**

5. **Anomaly Detection:**
- **Anomaly detection techniques identify data points that deviate from the norm or expected behavior. These outliers often represent patterns that are different from the majority of the data.**
- **One-class SVM, Isolation Forest, and Local Outlier Factor (LOF) are examples of algorithms used for anomaly detection.**

### **General Process of Unsupervised Learning**
1. **Unlabeled Input Data:**
    - The process starts with a dataset that lacks pre-assigned labels or categories. This raw data could be in various forms such as numerical values, images, text documents, etc.
    - Without labels, the algorithm doesn't have explicit guidance on how to categorize or interpret the data.

2. **Finding Hidden Patterns:**
    - The algorithm analyzes the raw data to identify any underlying patterns, structures, or relationships.
    - This involves exploring the statistical properties of the data to uncover similarities, differences, clusters, or other meaningful groupings.
    - The goal is to extract relevant information from the data without being explicitly told what to look for.

3. **Applying Suitable Algorithms:**
    - Once the algorithm has interpreted the raw data and identified potential patterns, it selects and applies appropriate unsupervised learning algorithms.
    - These algorithms are designed to categorize or organize the data based on its inherent characteristics. For example, clustering algorithms like **k-means** or **hierarchical clustering** group similar data points together, while dimensionality reduction techniques like **PCA** or **t-SNE** capture the essential features of the data.

4. **Dividing Data Objects into Groups:**
    - The selected algorithm partitions the data objects into groups or clusters based on similarities and differences among them.
    - Each group contains data points that are more similar to each other compared to data points in other groups.
    - The algorithm's objective is to minimize intra-group differences while maximizing inter-group differences, resulting in meaningful and coherent clusters.

# **Clustering**
- **Clustering algorithms group similar data points together based on certain criteria, such as distance or density.**

### **When to use clustering (Applications of clustering)?**
1. **Customer Segmentation:**
    - Clustering is widely used in marketing for segmenting customers into groups based on their purchasing behaviors, preferences, demographics, or other relevant features. These customer segments can then be targeted with tailored marketing strategies or product recommendations.

2. **Anomaly Detection:**
    - Clustering can be used for anomaly detection by identifying data points that deviate significantly from the norm or expected behavior. Outlier detection techniques based on clustering help detect unusual patterns or anomalies in various applications, such as fraud detection, network security, and fault diagnosis.

### **When not to use clustering?**
1. **No Natural Clusters:**
    - If the underlying data does not contain any meaningful clusters or if the clusters are not well-defined, clustering algorithms may struggle to produce meaningful results.

2. **High Dimensionality:**
    - Clustering algorithms may face challenges when dealing with high-dimensional data, where the distance metrics become less reliable due to the curse of dimensionality.

### **Can clustering be used to improve supervised machine learning algorithms?**
- Yes, clustering can be used to improve supervised machine learning algorithms in various ways. Here are some common approaches:
    1. Clustering can help preprocess the data before applying supervised learning algorithms. For instance, outlier detection techniques based on clustering can identify and remove noisy data points or correct mislabeled instances, leading to cleaner training data and better model performance.
    
    2. we can use unsupervised techniques as clustering to enhance our supervised models by clustering data points of datasets to separate clusters and use the cluster labels as independent variables in the supervised machine learning algorithm.

### **Types of Clustering Algorithms**
1. **Centroid-based Clustering:**
    
    Centroid-based is the one we mostly deal with where each data point is assigned to the cluster where it has the closest distance to its centroid.
    
    However such algorithms needs predefinition of the number of clusters which means that it needs prior information, before applying the algorithm, about the clusters that might exist in such data.

    This algorithm runs iteratively to find local optima and it is fast and efficient (K-means clustering).


2. **Hierarchical Clustering:**

    Hierarchical clustering builds a hierarchy of clusters, either from individual data points or from pre-existing clusters, based on the similarity between data points.

    Unlike centroid-based clustering, hierarchical clustering does not require specifying the number of clusters beforehand. Instead, the hierarchy can be explored at different levels to obtain clusters of varying granularity.

    There are two main approaches to hierarchical clustering:
    
    - **Agglomerative Hierarchical Clustering:** This bottom-up approach starts with each data point as a separate cluster and merges the closest clusters iteratively until all data points belong to a single cluster. It produces a dendrogram that shows the hierarchical structure of the clusters.

    - **Divisive Hierarchical Clustering:** This top-down approach starts with all data points in one cluster and recursively splits clusters until each data point is in its own cluster.


3. **Density-Based Clustering:**

    Density-based clustering algorithms operate under the assumption that clusters are regions of high density separated by regions of low density in the data space.

    Unlike centroid-based or hierarchical clustering, density-based clustering does not require specifying the number of clusters beforehand.

    Density-based clustering algorithms assign each data point to one of three categories: core points, border points, or noise points (outliers), based on their density and proximity to other points.

    DBSCAN is one of the most widely used density-based clustering algorithms.


### **Centroid-based Clusering (K-Means)**
- **K-means clustering is one of the most widely used unsupervised machine learning algorithms for partitioning data into K distinct, non-overlapping clusters.**

- **It's a centroid-based algorithm, meaning it treats each cluster as a centroid (or center point) and assigns data points to the nearest centroid.**

#### **How K-Means Works?**
- **Initialization:**
    - **K-means starts by randomly initializing K cluster centroids in the feature space. These centroids can be randomly selected data points or randomly generated coordinates within the range of the data.**

- **Assignment Step:**
    - **Each data point is then assigned to the nearest centroid based on a distance metric, typically the Euclidean distance. The data points are reassigned to the cluster with the closest centroid.**

- **Update Step:**
    - **After all data points are assigned to clusters, the centroids are recalculated as the mean (average) of all data points assigned to each cluster. This step updates the centroids to be the center points of their respective clusters.**

- **Convergence:**
    - **Steps 2 and 3 are repeated iteratively until either the centroids no longer change significantly between iterations or a specified number of iterations is reached. At this point, K-means is said to have converged.**

- **Final Result:**
    - **The final result of K-means clustering is K clusters, each represented by its centroid, and each data point belongs to the cluster associated with the nearest centroid.**


## Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from category_encoders import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import warnings
warnings.filterwarnings('ignore')
```

## Reading & Exploring Data

```python
df=pd.read_csv("Mall_Customers.csv")
df.shape
```

```python
def data_info(data):

    """
    This function returns a DataFrame containing the summary information for each column
    """

    Names=[col for col in data]
    data_types=[data[col].dtype for col in data.columns]
    top_10_unique_values=[data[col].value_counts().head(10).index.to_list() for col in data.columns]
    nunique_values=[data[col].nunique() for col in data.columns]
    nulls=[data[col].isnull().sum() for col in data.columns]
    percent_of_Nulls= [data[col].isnull().sum()/len(data)*100 for col in data.columns]
    duplicates=data.duplicated().sum()


    info_df=pd.DataFrame({'Name':Names,
                          'Data_Type':data_types,
                          'Top_10_Unique_Values':top_10_unique_values,
                          'Nunique_Values':nunique_values,
                          'Nulls':nulls,
                          'Percent_of_Nulls':percent_of_Nulls,
                          'Duplicates':duplicates})
    return info_df
```

```python
data_info(df)
```

```python
df.drop('CustomerID',axis=1,inplace=True)
```

```python
data_info(df)
```

## Encoding

```python
encoder = OneHotEncoder(cols=["Gender"], drop_invariant=False, use_cat_names=True)
df = encoder.fit_transform(df)

data_info(df)
```

## Scaling

```python
scaler=StandardScaler()

num_cols=["Age", "Annual Income (k$)", "Spending Score (1-100)"]
df[num_cols]=scaler.fit_transform(df[num_cols])

data_info(df)
```

## Choosing K Using Elbow

```python
wcss = []      # within-cluster sum of squares
K = range(1, 11)

for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(df)
    wcss.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K, wcss, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS (Inertia)')
plt.title('Elbow Method')
plt.show()
```

## Model Training

```python
kmeans = KMeans(n_clusters=4)
kmeans.fit_predict(df)

centroids = kmeans.cluster_centers_
labels = kmeans.labels_

print("Centroids:\n", centroids)
```

## Applying PCA for plots

```python
pca = PCA(n_components=2)
df_2d = pca.fit_transform(df)
centroids_2d = pca.transform(centroids)

plt.figure(figsize=(8,6))
plt.scatter(df_2d[:, 0], df_2d[:, 1], c=labels, s=40)

plt.scatter(centroids_2d[:, 0], centroids_2d[:, 1],
            marker='X', s=200, color='red')

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("KMeans Clusters (PCA Visualization)")
plt.show()
```

### **Advantages**
- **Efficiency:** K-Means is computationally efficient and scales well to large datasets, making it suitable for clustering large datasets with many data points.

- **Simple and Easy to Implement:** K-Means is straightforward to understand and implement, making it accessible even to users with limited machine learning experience.

- **Interpretability of Results:** The resulting clusters from K-Means are easy to interpret and visualize, especially in low-dimensional space. The cluster centroids serve as prototypes, representing the average of data points assigned to each cluster.

- **Scalability:** K-Means can handle datasets with a large number of clusters and is suitable for situations where the number of clusters is known or can be estimated.

- **Convergence Guarantee:** K-Means is guaranteed to converge to a local optimum, although the solution obtained depends on the initialization.


### **Disadvantages**
- **Sensitivity to Initialization:** K-Means clustering results can vary depending on the initial centroids' selection, leading to different cluster assignments and possibly suboptimal solutions. Poor initialization may result in convergence to a local minimum instead of the global minimum.

- **Assumption of Spherical Clusters:** K-Means assumes that clusters are spherical and have equal variance in all dimensions, which may not hold for all datasets. It may perform poorly on non-convex or irregularly shaped clusters.

- **Dependency on Number of Clusters (K):** The number of clusters (K) needs to be specified a priori, which can be challenging and may require domain knowledge or trial-and-error. Selecting an inappropriate number of clusters can lead to suboptimal results or misinterpretation of the data.

- **Sensitive to Outliers:** K-Means is sensitive to outliers, as they can significantly affect the centroids' positions and cluster assignments. Outliers may lead to the formation of erroneous clusters or distort the cluster boundaries.

- **Non-Robust to Noise:** K-Means clustering is non-robust to noise and may produce suboptimal results when the dataset contains noisy or irrelevant features.
